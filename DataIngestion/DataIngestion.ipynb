Langchain Document Loaders explanation
https://docs.langchain.com/oss/python/integrations/document_loaders

```python
## Text Loader
from langchain_community.document_loaders import TextLoader
LoaddedTextObject = TextLoader('speech.txt')
TextDoc = LoaddedTextObject.load()
TextDoc
```

```python
## PDF Loader
from langchain_community.document_loaders import PyPDFLoader
LoaddedPdfObject = PyPDFLoader('attention.pdf')
PdfDoc = LoaddedPdfObject.load()
PdfDoc
```

```python
## Web based Loader
from langchain_community.document_loaders import WebBaseLoader
## other parameters for playing with web scraping
LoaddedPdfObject = WebBaseLoader(web_paths=('https://docs.langchain.com/oss/python/integrations/document_loaders',),)
UrlDoc = LoaddedPdfObject.load()
UrlDoc

```
