Text Splittinmg From Documents

How to recursively splitting document(text) by characters

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
## PDF Loader
loadded_pdf_handler = PyPDFLoader('attention.pdf')
pdf_document = loadded_pdf_handler.load()
## Text Splitter
text_splitter_handler = RecursiveCharacterTextSplitter(chunk_size=100,chunk_overlap=50)
split_docs = text_splitter_handler.split_documents(pdf_document)
print(split_docs[0])
print(split_docs[1])
```
