Combine all steps of RAG 
1- load text (Data ingestion)
2- split text (Data transformation)
3- create embeddings using OpenAI API keys
4- store in vector db

```python
## Use OpenAi api keys for embedding
import os
from dotenv import load_dotenv
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
## load environment variables
load_dotenv() 
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
## Text Loader
LoaddedTextObject = TextLoader('speech.txt')
TextDoc = LoaddedTextObject.load()
## Text Splitter
text_splitter_handler = RecursiveCharacterTextSplitter(chunk_size=100,chunk_overlap=50)
split_docs = text_splitter_handler.split_documents(TextDoc)
embedding_handle = OpenAIEmbeddings(model="text-embedding-3-large")
## Create Embeddings and store in vector db (chromadb)
vector_db = Chroma.from_documents(split_docs, embedding_handle)
## Create query (request for similar text)
query_text = "it is the only thing we can never get back once it is gone"
## Response from vector db
retrived_result = vector_db.similarity_search(query_text)
print(retrived_result)
```
