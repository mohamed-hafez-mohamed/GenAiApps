```python
import os
from dotenv import load_dotenv
load_dotenv() # load all environment variables
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings
embedding_handle = OpenAIEmbeddings(model="text-embedding-3-large") # you can specifiy vector diemensions as second parameter
```

```python
text = "This is a tutorial OpenAi embeddings"
query_results = embedding_handle.embed_query(text)
query_results
```
