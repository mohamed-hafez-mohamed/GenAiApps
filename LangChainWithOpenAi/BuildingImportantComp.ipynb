```python
# Chat Prompt Template
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_messages(
    [
        # System message: tell the model its role
        ("system", "You are a helpful assistant that translates English to French."),
        # Human message: the user input
        ("user", "{input}")
    ]
)

```

```python
## Combine LLM and Prompt Template to create a chain
chain = prompt | llm
response = chain.invoke({"input": "What langsmith used for?"})
print(response)
```

```python
## stroutput parser to display message come from LLM
from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
chain = prompt | llm | output_parser
response = chain.invoke({"input": "What langsmith used for?"})
print(response)
```
